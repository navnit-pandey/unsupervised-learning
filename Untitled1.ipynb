```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

data = {
    'Student' : ['Amit', 'Priya','Rahul', 'Sneha','Karan','Neha',],
    'Math_Score' : [85,60,90,45,70,95],
    'Study_Hours' : [6,3,7,2,4,8]
}

df = pd.DataFrame(data)
print("Original Student Data: \n",df)

features = ['Math_Score', 'Study_Hours']
x = df[features]

scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

df_scaled=pd.DataFrame(x_scaled, columns=features)
df_scaled.insert(0, 'Student', df['Student'])
print("\nScaled Student Data: \n",df_scaled)

plt.figure(figsize=(10, 6))
plt.scatter(df['Math_Score'], df['Study_Hours'], color='blue', s=80, label='Before Scaling')
plt.scatter(df_scaled['Math_Score'], df_scaled['Study_Hours'], color='red', s=80, label='After Scaling')
plt.title('Math Score vs. Study Hours Before and After Scaling')
plt.xlabel('Math Score')
plt.ylabel('Study Hours')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
pd.plotting.scatter_matrix(x, figsize=(10, 8))
plt.show()
```

```python
from sklearn.cluster import KMeans

# Initialize and fit the KMeans model
KMEAN = KMeans(n_clusters=3, random_state=42, n_init=10) # Assuming 3 clusters for demonstration
KMEAN.fit(x_scaled)

print("Cluster labels:", KMEAN.labels_)
print("Cluster centers:\n", KMEAN.cluster_centers_)
```

```python
plt.figure(figsize=(10, 5))
plt.scatter(x_scaled[:,0],x_scaled[:,1],c = KMEAN.labels_,marker="x",label = "datapoint")
plt.xlabel("annual income")
plt.ylabel("spending score")
plt.title("Clusters")
plt.scatter(
    KMEAN.cluster_centers_[:, 0],
    KMEAN.cluster_centers_[:, 1],
    marker = "*",
    s=300,
    c="red",
    label="centroids",
)
plt.legend()
plt.show()
```
