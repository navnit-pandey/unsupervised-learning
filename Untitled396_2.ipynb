```python
import pandas as pd
import matplotlib.pyplot as plt
data = {"Annual Income": [15,16,17,18,19,60,62,65,67,70],
        "Spending Score": [39,42,45,40,43,70,65,72,68,75]
        }
df = pd.DataFrame(data)
df.head()
```

```python
from sklearn_extra.cluster import KMedoids
kmedoids = KMedoids(n_clusters=2, random_state=42)
model = kmedoids.fit(df)
df['Cluster'] = kmedoids.labels_
print(df)
print("Mediods")
print(df.iloc[kmedoids.medoid_indices_])
```
