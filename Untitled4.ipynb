```python
from sklearn.cluster import MiniBatchKMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
X, _= make_blobs(n_samples = 1000, centers = 3, n_features=2,random_state = 42)
mbk_means = MiniBatchKMeans(n_clusters=3, random_state=42)
mbk_means.fit(X)

plt.scatter(X[:,0], X[:,1], c=mbk_means.labels_ ,cmap='viridis')
plt.scatter(mbk_means.cluster_centers_[:,0], mbk_means.cluster_centers_[:,1], s=200, c='red', marker='x')
plt.title('Mini-Batch K-Means Clustering')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
data  = pd.read_csv('Iris.csv')
X = data[["SepalLengthCm", "PetalLengthCm"]]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
kmeans = KMeans(n_clusters=5, init = 'k-means++', random_state=42, n_init=10) # Corrected KMeans capitalization, added missing comma, and n_init
kmeans.fit(X_scaled)
data['cluster'] = kmeans.labels_ # Corrected 'cluster' to a string
print("Cluster Centers:")
print(kmeans.cluster_centers_)
print(data.head())

plt.figure(figsize=(8, 6))
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=kmeans.labels_, cmap='viridis', s=50, alpha=0.8)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=200, c='red', marker='X', label='Centroids')
plt.title('K-Means Clustering on Iris Dataset (Scaled Features)')
plt.xlabel('Scaled Sepal Length')
plt.ylabel('Scaled Petal Length')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
